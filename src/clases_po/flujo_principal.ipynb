Primero cargué el dataset iris de Hugging Face con mi primera clase. Después hice una transformación: filtré el DataFrame y me quedé solo con las filas donde SepalLengthCm es mayor a 5. Por último, le pasé ese DataFrame filtrado a mi segunda clase, que generó el resumen en un archivo .txt

```python
from hf_dataset_loader import HFDatasetLoader
from dataset_summarizer import DatasetSummarizer
```

```python
loader = HFDatasetLoader("scikit-learn/iris", split="train")
df = loader.load()
print("Filas y columnas originales:", df.shape)
print(df.columns.tolist())
df.head()
```

```python
df_transformado = df[df["SepalLengthCm"] > 5].copy()
print("Filas después de filtrar:", df_transformado.shape)
df_transformado.head()
```

```python
resumidor = DatasetSummarizer(df_transformado, nombre="iris (filtrado: SepalLengthCm > 5)")
ruta = resumidor.guardar_txt("resumen_iris_transformado.txt")
print("Archivo generado:", ruta)
```

```python
with open("resumen_iris_transformado.txt", encoding="utf-8") as f:
    print(f.read())
```

Descargué a mano un CSV (iris) y un JSON de Hugging Face y guardé los originales sin tocarlos. Con la clase ManejadorCSV cambié el valor Iris-setosa por setosa y eliminé un registro, y guardé un CSV nuevo. Con ManejadorJSON actualicé un campo y agregué un registro, y guardé un JSON nuevo. Al final, la clase GeneradorResumenTXT recibió los resultados de las dos clases y creó resumen.txt con las URL, las modificaciones, las rutas y la cantidad de registros

```python
url_csv = "https://huggingface.co/datasets/scikit-learn/iris/blob/main/Iris.csv"
url_json = "https://huggingface.co/datasets/geekyfreak/json_dump/blob/main/data.json"

ruta_csv_original = "datos_originales/Iris.csv"
ruta_json_original = "datos_originales/data.json"
ruta_csv_nuevo = "datos_modificados/iris_modificado.csv"
ruta_json_nuevo = "datos_modificados/data_modificado.json"
```

```python
from manejador_csv import ManejadorCSV
from manejador_json import ManejadorJSON
from generador_resumen_txt import GeneradorResumenTXT

manejador_csv = ManejadorCSV(ruta_csv_original)
manejador_json = ManejadorJSON(ruta_json_original)
```

```python
encabezados, filas = manejador_csv.importar()
print("Encabezados:", encabezados)
print("Registros importados:", len(manejador_csv.filas))

antes = [fila[:] for fila in manejador_csv.filas[:3]]
print("\nANTES (primeras 3 filas):")
for fila in antes:
    print(fila)
```

```python
manejador_csv.actualizar_valor("Species", "Iris-setosa", "setosa")
manejador_csv.eliminar_registro("Id", "150")

print("DESPUÉS (primeras 3 filas):")
for fila in manejador_csv.filas[:3]:
    print(fila)
print("Registros ahora:", len(manejador_csv.filas))
```

```python
ruta = manejador_csv.guardar(ruta_csv_nuevo)
print("CSV nuevo guardado en:", ruta)

with open(ruta, "r", encoding="utf-8") as f:
    for _ in range(4):
        print(f.readline().strip())
```

```python
registros = manejador_json.importar()
print("Registros importados:", len(registros))
print("Campos del primer registro:", list(registros[0].keys()))
print("category ANTES:", repr(registros[0].get("category")))
```

```python
manejador_json.actualizar_valor(0, "category", "facturas")
manejador_json.agregar_registro({
    "context": "Registro agregado desde Jupyter",
    "output": [],
    "category": "nuevo"
})

print("category DESPUÉS:", repr(manejador_json.registros[0].get("category")))
print("Registros ahora:", len(manejador_json.registros))
```

```python
ruta = manejador_json.guardar(ruta_json_nuevo)
print("JSON nuevo guardado en:", ruta)

import json
with open(ruta, "r", encoding="utf-8") as f:
    verificacion = json.load(f)
print("Registros en el archivo nuevo:", len(verificacion))
print("Último registro:", verificacion[-1]["category"])
```

```python
generador = GeneradorResumenTXT(
    url_csv,
    url_json,
    manejador_csv.resultados(),
    manejador_json.resultados(),
)
ruta_resumen = generador.generar_txt("resumen.txt")
print("Resumen generado en:", ruta_resumen, "\n")

with open(ruta_resumen, "r", encoding="utf-8") as f:
    print(f.read())
```
